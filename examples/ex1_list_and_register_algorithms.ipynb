# MAAP Client Library Examples

Demonstrates instantiation, algorithms service, and job submission.

```python
from maap import MAAP

maap = MAAP(host="https://api.uat.maap-project.org")
```

```python
algorithms = maap.algorithms.list()

print(f"Total algorithms: {len(algorithms)}\n")
for algo in algorithms[:5]:
    print(f"{algo.process_id:>4}  {algo.title} (v{algo.version}) — {algo.author}")
```

```python
algo = maap.algorithms.get(algorithm_id="96")


print(f"Title:       {algo.title}")
print(f"Author:      {algo.author}")
print(f"Deployed by: {algo.deployed_by}")
print(f"RAM min:     {algo.ram_min} GB")
print(f"Cores min:   {algo.cores_min}")
print(f"Inputs:      {list(algo.inputs.keys())}")
```

## Job Submission

Build a `JobRequest` and submit it through the jobs service.

```python
from maap.types import JobRequest

request = JobRequest(algorithm_id=96, inputs={"SHORT_NAME": "test", "TEMPORAL": "2024-01-01/2024-12-31"})
job = maap.jobs.submit(request)

print(f"Job ID:  {job.job_id}")
print(f"Status:  {job.status}")
print(f"Created: {job.created}")
```
